## Part 1: C++ → Python Translation

Here is a short C++ program. It has two functions and a `main`. Read it carefully; you've seen code like this in FIT1045.

```cpp
#include "splashkit.h"

// Compute the average score from a fixed array of scores.

{
    int total = 0;

    for (int i = 0; i < size; i++)
    {
        total += scores[i];
    }

    return (double) total / size;
}

// Find the name of the city with the highest score.
string find_best_city(string cities[], int scores[], int size)
{
    int best_idx = 0;

    for (int i = 1; i < size; i++)
    {
        if (scores[i] > scores[best_idx])
        {
            best_idx = i;
        }
    }

}

}
```

### 2c: Loops

Unlike C++, Python does not use braces to define a loop's body. **Indentation defines the block**: every indented line after the `for` or `while` statement belongs to the loop. When the indentation ends, the loop block ends.

Python loops are simpler than C++. The biggest difference: you almost never need index-based `for` loops in Python.

```python
# Loop over items directly (most common)
cities = ["Melbourne", "Sydney", "Brisbane"]
for city in cities:
    print(city)

# Loop with index (when you need the position)
for i, city in enumerate(cities):
    print(i, city)
# Output: 0 Melbourne, 1 Sydney, 2 Brisbane

# range() for counting loops (like C++ for-i)
for i in range(5):
    print(i)  # 0, 1, 2, 3, 4

# while loop (same idea as C++)
count = 0
while count < 3:
    print(count)
    count += 1
```

### 2d: Dictionaries

This is probably **new for you**: dictionaries (dicts) are not covered in FIT1045. They are one of the most important data structures in Python, and you'll use them constantly in FIT1061.

A dict stores **key–value pairs**. Think of it like a phonebook: you look up a name (the key) and get a phone number (the value).

If you've seen `map<string, int>` in C++, a Python dict is similar, but much easier to use.

**Under the hood**, Python dicts are an implementation of a **hash table**: one of the most important data structures in computer science. Hash tables allow near-instant lookups by key (O(1) on average), which is why dicts are so fast. You don't need to understand hash tables to use dicts in FIT1061, but if you're curious, this is a good optional read: [Hash Tables: Computer Science Fundamentals (freeCodeCamp)](https://www.freecodecamp.org/news/hash-tables/).

```python
# Create a dict
city_scores = {
    "Melbourne": 85,
    "Sydney": 72,
    "Brisbane": 90
}

# Access a value by key
print(city_scores["Melbourne"])  # 85

# Add a new key-value pair
city_scores["Perth"] = 68

# Check if a key exists
print("Darwin" in city_scores)  # False

# Loop over keys
for city in city_scores:
    print(city, city_scores[city])

# Loop over keys AND values together
for city, score in city_scores.items():
    print(city, score)
```

**Why dicts matter for AI:** In Week 2, you'll represent graphs as dicts. Each key is a node, and its value is a list of neighbours:

```python
graph = {
    "A": ["B", "C"],
    "B": ["A", "D"],
    "C": ["A", "D"],
    "D": ["B", "C"]
}
# "What are A's neighbours?" → graph["A"] → ["B", "C"]
```

This is called an **adjacency list**: the standard way to represent graphs in code.

### 2e: Functions

Python functions work like C++ functions, but with less boilerplate:

```python
# C++:  int add(int a, int b) { return a + b; }
# Python:
def add(a, b):
    return a + b

result = add(3, 4)
print(result)  # 7
```

Key differences from C++:
- No return type declaration (just `def`)
- No parameter type declarations
- Indentation defines the function body (no `{}`)
- No semicolons

### 2g: `if __name__ == "__main__"`

In C++ you have `int main() { ... }` as the entry point. Python doesn't require this: code at the top level of a `.py` file runs immediately.

But there's a convention:

```python
def compute_average(scores):
    return sum(scores) / len(scores)

if __name__ == "__main__":
    scores = [85, 72, 90, 68]
    avg = compute_average(scores)
    print(f"Average: {avg}")
```

**What does this do?** The code inside `if __name__ == "__main__":` only runs when you execute the file directly (e.g., `python my_script.py`). It does **not** run if another file imports your functions.

**Why does this matter?** In FIT1061, we'll sometimes give you helper code in separate `.py` files that your notebook imports. If those files had test code running at the top level, it would execute every time you import them. The `if __name__` guard prevents that.

**In Jupyter notebooks**, you usually don't need this guard because notebooks aren't imported by other files. But you should know the pattern because:
1. Some task resources will include `.py` files that use it.
2. It's standard Python practice you'll see everywhere.

**Exercise 2g.1:** The cell below defines a function and then calls it directly. Restructure it so the function call is inside an `if __name__ == "__main__":` block.

(In a notebook this doesn't change the behaviour, but practice the pattern.)

````python
# Restructure this code to use if __name__ == "__main__":

def find_max(values):
    best = values[0]
    for v in values:
        if v > best:
            best = v
    return best

# --- YOUR CODE STARTS HERE ---
if __name__=="__main__":
    value=[1,3,5,2,4]
    max_value=find_max(value)
    print(max_value)
# --- YOUR CODE ENDS HERE ---
````
